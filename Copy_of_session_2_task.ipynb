### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
salary = int(input("Enter the annual salary: "))
hra = 10/100
da = 5/100
pf = 3/100
fixed = salary * (hra + da + pf)
if salary < 500000:
  total = 0
elif salary >= 500000 and salary < 1000000:
  total = (salary-500000) * 10/100
elif salary >= 1000000 and salary < 2000000:
  total = 500000*(10/100) - (salary-1000000) * (20/100)
else:
  total = 500000*(10/100) + 1000000*(20/100) - (salary - 2000000)*(30/100)
final_salary = salary - total - fixed
final_salary = final_salary/12
print(f"Your monthly in hand salary is {final_salary}")
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# Write code here
fact_num = int(input("Enter the number you want a factorial of: "))
i = 1
fact = 1
while i <= fact_num:
  fact = fact * i
  i = i + 1
print(fact)
````

### `Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

```bash

```

```bash
276
```

````python
# Write code here
n = int(input("Enter the number "))
i = 1
sum = 0
while i <= n:
  if i%5 == 0:
    i = i + 1
    continue
  else:
    sum = sum + i
    if sum > 300:
      break
  i = i + 1
print(sum)




````

### `Problem 9`: Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

````python
# Write code here
sum = 0
count = 0

while True:
    num = int(input("Enter a number (0 to stop): "))

    if num == 0:
        break

    sum += num
    count += 1

if count > 0:
    average = sum / count
else:
    average = 0

print(f"Sum of numbers: {sum}")
print(f"Average of numbers: {average}")
````

###`Problem 9`: Write a program which will find all such numbers which are divisible by 7 but are not a multiple of 5, between 2000 and 3200 (both included). The numbers obtained should be printed in a comma-separated sequence on a single line.

````python
# Write code here
result = []

for number in range(2000, 3201):  # 3201 because range upper bound is exclusive
    if number % 7 == 0 and number % 5 != 0:
        result.append(str(number))  # Convert number to string for joining later

print(",".join(result))
````

###`Problem 10`: Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number. The numbers obtained should be printed in a space-separated sequence on a single line.



````python
# Write code here
result = []

for number in range(1000, 3001):
    num_str = str(number)

    if (int(num_str[0]) % 2 == 0 and
        int(num_str[1]) % 2 == 0 and
        int(num_str[2]) % 2 == 0 and
        int(num_str[3]) % 2 == 0):

        result.append(num_str)

print(" ".join(result))
````

###`Problem 12`:Write a program to print whether a given number is a prime number or not

````python
# Write code here
num = int(input("Enter a number: "))

if num <= 1:
    print("Not a prime number")
else:
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime number")
    else:
        print("Not a prime number")
````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
# Write code here
a_num = int(input("Enter the number: "))
l = a_num
sum= 0
while a_num > 0:
  num = a_num%10
  sum = sum + num**3
  a_num = a_num//10
if l == sum:
  print("Number is an Armstrong number")
else:
  print("Number is not an Armstrong number")
````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
# Write code here
H = int(input("Enter hour (0-12): "))
M = int(input("Enter minutes (0-59): "))

# Calculate positions
hour_angle = H * 30 + M * 0.5
minute_angle = M * 6

# Find the absolute difference
angle = abs(hour_angle - minute_angle)

# Minimum of the two possible angles
min_angle = min(angle, 360 - angle)

# Floor the result
print(int(min_angle))
````
